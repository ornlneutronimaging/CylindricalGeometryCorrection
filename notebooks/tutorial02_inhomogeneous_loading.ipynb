```python
from neutron_geomcorr.geometry_correction import GeometryCorrection
import glob
```

# Homogeneous 

## tiff images 

```python
list_tiff = glob.glob('data/tiff/inhomogeneous*.tif')

# initialize object
o_cgc = GeometryCorrection(list_files=list_tiff)
o_cgc.load_files(notebook=True)

# center of cylinder
pixel_center = 256

# inner radius of cylinder
inner_radius = 150

# outer radius of cylinder
outer_radius = 200

# define the parameters
o_cgc.define_parameters(inner_radius=inner_radius, outer_radius=outer_radius, pixel_center=pixel_center)
```

## fits images

```python
list_fits = glob.glob('data/fits/inhomogeneous*.fits')

o_cgc = GeometryCorrection(list_files=list_fits)
o_cgc.load_files(notebook=True)
```
