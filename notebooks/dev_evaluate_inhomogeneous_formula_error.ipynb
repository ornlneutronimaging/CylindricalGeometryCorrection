# Instructions 

This notebook will load real fits cylindrical sample images with various Intensity/radius to evaluate the difference between the output given by the formula and the real data.

# Python Import

```python
from plotly.offline import plot, init_notebook_mode, iplot
init_notebook_mode()
import plotly.graph_objects as go
from plotly import tools

import numpy as np
import os

from IPython.display import Image
from IPython.core.display import HTML

import matplotlib.pyplot as plt
%matplotlib inline

from neutron_geomcorr import file_handler
from neutron_geomcorr import geometry_correction
```

# Working with Inomogeneous Image

## Load and Preview of data

```python
list_files = ['data/fits/inhomogeneous_image_px_intensity_2.fits',
              'data/fits/inhomogeneous_image_px_intensity_4.fits',
              'data/fits/inhomogeneous_image_px_intensity_6.fits',
              'data/fits/inhomogeneous_image_px_intensity_8.fits']

list_intensity = []
for _file in list_files:
    [_name, _] = os.path.splitext(_file)
    _intensity = os.path.basename(_name).split('_')[-1]
    list_intensity.append(_intensity)

inhomogeneous_image = []
for _file in list_files:
    inhomogeneous_image.append(file_handler.load_fits(_file))
    
```

## Retrieve/Define Data Characteristics and Correct Geometry

```python
expected_slice_array = []
pixel_intensity_array = []
for _image in inhomogeneous_image:

    #image size
    [width, height] = np.shape(_image)

    # defined by user
    pixel_center = 256
    inner_radius = 150
    outer_radius = 200

    _projected_image = _image[0,:]
    pixel_intensity = _projected_image[pixel_center]/(2*(outer_radius-inner_radius))
    pixel_intensity_array.append(pixel_intensity)

    truncated_image = _image[:, pixel_center-outer_radius:pixel_center+outer_radius+1]
    [height, new_width] = np.shape(truncated_image)

    # work only on 1 slice
    expected_slice = []
    _h = 0 # slice index
    _slice = truncated_image[_h, :]
    for _index, _x in enumerate(np.arange(-outer_radius, outer_radius+1)):
        _input_data = _slice[_index]
        # Calculate the correction factor for inhomogeneous case
        _rp = geometry_correction.GeometryCorrection.inhomogeneous_correction(x=_x, inner_radius=inner_radius, outer_radius=outer_radius)
        if _rp == 0:
            _coeff = np.nan
        else:
            _coeff = (2*(outer_radius-inner_radius))/_rp
        _corrected_value = (_input_data * _coeff) / 2.0
        expected_slice.append(_corrected_value)
        
    expected_slice_array.append(expected_slice)
```

## Compare with real Results

```python
real_output = []
real_output.append(np.ones(len(expected_slice))*100)
real_output.append(np.ones(len(expected_slice))*200)
real_output.append(np.ones(len(expected_slice))*300)
real_output.append(np.ones(len(expected_slice))*400)

real_vs_expected = zip(real_output, expected_slice_array)

data = []
error_array = []
for _index, [_real, _expected] in enumerate(real_vs_expected):
    error_array.append(np.abs(_real-_expected))
    trace = go.Scatter(y=np.abs(_real[1:] - _expected[1:]),
                  name="Ipx: {}".format(list_intensity[_index]))
    data.append(trace)

figure1 = go.Figure(data=data)
iplot(figure1)
```

## Trying to Find Equation of Error Profile

```python
# keep only 1 side of symetrical error signal

right_side_error_array = []
error_data = []
for _index, _error in enumerate(error_array):
    _right_side_value = _error[int(len(_error)/2):]
    right_side_error_array.append(_right_side_value)
    trace = go.Scatter(y=_right_side_value,
                  name="Ipx: {}".format(list_intensity[_index]))
    error_data.append(trace)
    
figure2 = go.Figure(data=error_data)
iplot(figure2)
```

```python
# working with profile #: index
index = 3

profile = right_side_error_array[index]

r1 = 150
r2 = 200

coeff = 2*(r2-r1)* (pixel_intensity_array[index]/2)
d = np.arange(0,len(profile))
factor=[]
for _d in d:
    _result = geometry_correction.GeometryCorrection.inhomogeneous_correction(x=_d, inner_radius=r1, outer_radius=r2)
    if _result == 0:
        _result = np.nan
    else:
        _result = coeff / _result
    factor.append(_result)


error_data = go.Scatter(y=profile,
                        name='Error')
equation_data = go.Scatter(y=factor,
                          name='Equation')
data = [equation_data, error_data]
figure3 = go.Figure(data=data)
iplot(figure3)
```

## Plot Expected with Errors Bars

```python
index = 2

profile = right_side_error_array[index]

r1 = 200
d = np.arange(0,len(profile))
rp = 2*r1*np.sin(np.arccos(d/r1))
factor = (2*r1)/rp * (pixel_intensity_array[index]/2)
_error = np.concatenate((np.array(factor[::-1]), np.array(factor[1:-1])), axis=0)

_expected_plot = go.Scatter(y=expected_slice_array[index][1:],
                           name='expected',
                           mode='markers',
                           error_y = dict(type='data',
                                         array=_error,
                                         visible=True))


figure4 = go.Figure(data=[_expected_plot])
iplot(figure4)
```
